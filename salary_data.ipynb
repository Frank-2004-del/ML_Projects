```python
# Importing some libraries
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("Salary_dataset.csv")
```

```python
X = data["YearsExperience"]
y = data["Salary"]
plt.scatter(X,y)
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Scatter plot of data")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_tr,x_te,y_tr,y_te = train_test_split(X,y,test_size=0.3,random_state=32)
x_train = x_tr.to_numpy().reshape(-1,1)
y_train = y_tr.to_numpy().reshape(-1,1)
x_test = x_te.to_numpy().reshape(-1,1)
y_test = y_te.to_numpy().reshape(-1,1)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error, r2_score
y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)
mse_train = (mean_squared_error(y_train,y_train_pred))
mse_test = (mean_squared_error(y_test,y_test_pred))
print(f"Training mse:{mse_train}")
print(f"Test mse:{mse_test}")
r2_train = r2_score(y_train,y_train_pred)
r2_test = r2_score(y_test,y_test_pred)
print(f"Train r2score:{r2_train}")
print(f"Test_r2score:{r2_test}")
import numpy as np
rmse_test = np.sqrt(mse_test)
mae_test = mean_absolute_error(y_test,y_test_pred)
print(f"Test rmse is:{rmse_test}")
print(f"Test mae is: {mae_test}")
```

```python
plt.scatter(X,y)
plt.plot(x_train,y_train_pred,color="red")
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Full Visualization")
```
