```python
# importing required libraries
import pandas as pd
import numpy as np
```

```python
def train(x,y):
    """
    This function trains the simple linear regression model
    Input variables
    x: input features
    y: target of features
    return:
    w0 = bias or intercept of model
    w1 = slope
    """
    x_bar = np.mean(x)
    y_bar = np.mean(y)
    w1= np.sum((x-x_bar)*(y-y_bar))/np.sum((x-x_bar)**2)
    w0 = y_bar - w1*x_bar
    return w0,w1
```

```python
def predict(x,w0,w1):
    """
    This function gives prediction of an input
    Input variables
    x: Input feature
    w0,w1: Trained parameters of model
    return:
    y = prediction value
    """
    y = w0 + w1*x
    return y
    
```

```python
#evaluating performance by calculating mse
def error(x,y,w0,w1):
    """
    This function gives the error of the prediction
    Input:
    x,y: input features and target variable
    w0,w1: Trained parameters of model
    return:
    mse: mean squared error
    """
    mse = 0
    for i,j in zip(x,y):
        y_pred = predict(i,w0,w1)
        rss += (y_pred - j)**2
    mse = mse/len(x)
    return rss
 
```
