```python
# Importing libraries .
import pandas as pd
import numpy as np
```

```python
def train(x,y):
    """
    This function trains model to get parameters using closed form solution
    Input:
    x: input feature vector
    y: target vector
    return:
    w: trained weights of model
    """
    w = np.linalg.inv((x.T @ x)) @ x.T @ y
    return w
```

```python
def predict(x,w):
    """
    This function gives prediction of a feature
    Input:
    x: Input feature
    w: Trained parameters
    return:
    y: predicted target of feature x
    """
    y = x @ w
    return y
```

```python
def error(x,y,w):
    """
    This function calculates error metric of model
    input:
    x: input vector
    y: target values
    w: trained weights of model
    """
    mse = 0
    for i in range(len(x)):
        y_pred = predict(x[i],w)
        mse += (y_pred - y[i])**2
    mse = mse/len(x)
    mse
```
